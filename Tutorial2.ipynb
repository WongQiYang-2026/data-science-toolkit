```python
import numpy as np

array_2x2 = np.array ([
    [1,2],
    [3,4]
])

print("This is my numpy 2x2 array")
print(array_2x2)
```

```python
random3x4 = np.random.rand(3,4)

print("Random 3x4 array")
print(random3x4)
```

```python
ones_array = np.ones((3,3))
zeros_array = np.zeros((3,3))

print("Array of ones")
print(ones_array)

print("Array of zeros")
print(zeros_array)
```

```python
random6x3 = np.random.rand(6,3)

print(random6x3)

Trans = np.transpose(random6x3)

print("Transpose of Random6x3")
print(Trans)
```

```python
n=3
p=4
m=2

array_nxp = np.random.rand(n,p)
array_pxm = np.random.rand(p,m)

print("array_nxp:")
print(array_nxp )
print("array_pxm:")
print(array_pxm )

Dot_Product = np.dot(array_nxp,array_pxm)

print("Dot_Product:")
print(Dot_Product)

print("\nOption 2")
print(array_nxp@array_pxm)

print("\nShape", Dot_Product.shape)
```
